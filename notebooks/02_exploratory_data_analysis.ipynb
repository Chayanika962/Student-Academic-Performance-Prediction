## 02.EXPLORATORY DATA ANALYSIS(EDA)

In this section, we will explore the Student Performance dataset using statistical methods and visualizations.
We will perform univariate, bivariate and multivariate analysis to identify    patterns and relationships in the data.

```python
# Import libraries for data analysis and visualization
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the student performance dataset
df = pd.read_csv("../data/student-mat.csv", sep=";")
```

```python
# Create the target column using the final grade
df["Result"] = df["G3"].apply(lambda x: "Pass" if x >= 10 else "Fail")
```

```python
# Display the first five rows
df.head()
```

#UNIVARIATE ANALYSIS

```python
# Plot the distribution of students' ages
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="age", bins=10, kde=True)

plt.title("Distribution of Students' Age")
plt.xlabel("Age")
plt.ylabel("Number of Students")
plt.show()
```

```python
# Plot the number of male and female students
plt.figure(figsize=(6, 5))
sns.countplot(data=df, x="sex")

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Students")
plt.show()
```

```python
# Plot the number of students who passed and failed
plt.figure(figsize=(6, 5))
sns.countplot(data=df, x="Result")

plt.title("Pass and Fail Distribution")
plt.xlabel("Result")
plt.ylabel("Number of Students")
plt.show()
```

```python
# Plot the distribution of weekly study time
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="studytime")

plt.title("Distribution of Weekly Study Time")
plt.xlabel("Study Time Category")
plt.ylabel("Number of Students")
plt.show()
```

#BIVARIATE ANALYSIS

```python
# Compare study time with pass and fail results
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="studytime", hue="Result")

plt.title("Study Time vs Student Result")
plt.xlabel("Study Time Category")
plt.ylabel("Number of Students")
plt.show()
```

```python
# Compare previous failures with student results
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="failures", hue="Result")

plt.title("Previous Failures vs Student Result")
plt.xlabel("Number of Previous Failures")
plt.ylabel("Number of Students")
plt.show()
```

```python
# Examine the relationship between absences and final grade
plt.figure(figsize=(7, 5))
sns.scatterplot(data=df, x="absences", y="G3", hue="Result")

plt.title("Absences vs Final Grade")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade")
plt.show()
```

#MULTIVARIATE ANALYSIS

```python
# Select numerical columns for correlation analysis
numeric_df = df.select_dtypes(include="number")

# Calculate correlation between numerical variables
correlation = numeric_df.corr()

# Display the correlation heatmap
plt.figure(figsize=(14, 10))
sns.heatmap(correlation, cmap="coolwarm", annot=False)

plt.title("Correlation Heatmap of Student Performance")
plt.show()
```
