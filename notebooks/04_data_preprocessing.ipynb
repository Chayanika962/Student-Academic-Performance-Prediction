## 04.DATA PREPROCESSING

In this section, we will prepare the student performance dataset for machine learning.

We will separate input features and target variables,split the dataset into training and testing sets,scale numerical features and encode categorical features.

```python
# Import libraries for data preprocessing
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
# Load the student performance dataset
df = pd.read_csv("../data/student-mat.csv", sep=";")

# Create the Pass/Fail target
df["Result"] = df["G3"].apply(
    lambda x: "Pass" if x >= 10 else "Fail"
)

# Display the first five rows
df.head()
```

```python
# Separate input features by removing grade columns
X = df.drop(columns=["G1", "G2", "G3", "Result"])

# Select the target variable
y = df["Result"]

# Display the input features and target
print("Input features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# Identify numerical columns
numeric_features = X.select_dtypes(include=["int64", "float64"]).columns

# Identify categorical columns
categorical_features = X.select_dtypes(include=["object","str"]).columns

# Display the column names
print("Numerical columns:")
print(list(numeric_features))

print("\nCategorical columns:")
print(list(categorical_features))
```

```python
# Divide the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Display the shapes of the datasets
print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
# Check Pass and Fail distribution in training data
print("Training data distribution:")
print(y_train.value_counts(normalize=True) * 100)

# Check Pass and Fail distribution in testing data
print("\nTesting data distribution:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
# Create the preprocessing transformer
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)

print("Preprocessing transformer created successfully.")
```

```python
# Fit preprocessing on training data and transform it
X_train_processed = preprocessor.fit_transform(X_train)

# Transform testing data using the same fitted preprocessing
X_test_processed = preprocessor.transform(X_test)

# Display the processed data dimensions
print("Processed training data:", X_train_processed.shape)
print("Processed testing data:", X_test_processed.shape)
```
