## 03.HYPOTHESIS TESTING

In this section, we will perform statistical hypothesis tests to investigate the relationships between student academic performance and selected variables.
We will use a significance level of 0.05.

```python
# Import libraries for hypothesis testing
import pandas as pd
from scipy.stats import chi2_contingency
from scipy.stats import ttest_ind
```

```python
# Load the student performance dataset
df = pd.read_csv("../data/student-mat.csv", sep=";")

# Create the Pass/Fail target
df["Result"] = df["G3"].apply(lambda x: "Pass" if x >= 10 else "Fail")

# Set the significance level
alpha = 0.05
```

#HYPOTHESIS TEST 1: STUDY TIME vs RESULT

```python
# Create a contingency table
table1 = pd.crosstab(df["studytime"], df["Result"])

# Perform the Chi-square test
chi2, p_value, dof, expected = chi2_contingency(table1)

# Display the results
print("Chi-square statistic:", chi2)
print("P-value:", p_value)

# Make the decision
if p_value <= alpha:
    print("Reject H0: Study time is significantly associated with Result.")
else:
    print("Fail to reject H0: No significant association was found.")
```

#HYPOTHESIS TEST 2: ABSENCES vs RESULT

```python
# Separate absences according to student result
pass_absences = df[df["Result"] == "Pass"]["absences"]
fail_absences = df[df["Result"] == "Fail"]["absences"]

# Calculate the average absences
print("Average absences of Pass students:", pass_absences.mean())
print("Average absences of Fail students:", fail_absences.mean())

# Perform Welch's independent samples t-test
t_stat, p_value = ttest_ind(
    pass_absences,
    fail_absences,
    equal_var=False
)

# Display the results
print("T-statistic:", t_stat)
print("P-value:", p_value)

# Make the decision
if p_value <= alpha:
    print("Reject H0: Mean absences differ significantly.")
else:
    print("Fail to reject H0: No significant difference was found.")
```

#HYPOTHESIS TEST 3: PREVIOUS FAILURES vs RESULT

```python
# Create a contingency table
table2 = pd.crosstab(df["failures"], df["Result"])

# Perform the Chi-square test
chi2, p_value, dof, expected = chi2_contingency(table2)

# Display the results
print("Chi-square statistic:", chi2)
print("P-value:", p_value)

# Make the decision
if p_value <= alpha:
    print("Reject H0: Previous failures are significantly associated with Result.")
else:
    print("Fail to reject H0: No significant association was found.")
```

```python
# Check expected frequencies for the Chi-square tests
print("Expected frequencies for Study Time:")
print(expected if False else "We will calculate these separately below.")

# Recalculate expected frequencies for Study Time
study_table = pd.crosstab(df["studytime"], df["Result"])
_, _, _, expected_study = chi2_contingency(study_table)

print(expected_study)

# Check expected frequencies for Previous Failures
failure_table = pd.crosstab(df["failures"], df["Result"])
_, _, _, expected_failures = chi2_contingency(failure_table)

print("\nExpected frequencies for Previous Failures:")
print(expected_failures)
```

## Conclusion of Hypothesis Testing

1. Study time and student results did not show a statistically
significant association.

2. The average absences of Pass and Fail students did not show
a statistically significant difference.

3. Previous failures showed a statistically significant
association with student results.

The expected-frequency checks for both Chi-square tests
satisfied the usual minimum-count guideline.
